```python
# Function to detect the count of outliers:
def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)
    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]
    print("Column:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))
    return outliers
```

```python
booking_value_outliers = detect_outliers_iqr(
    bookings,
    "Booking_Value"
)
```

## ANOVA:

```python
booking_vehicle = bookings.merge(
    vehicles,
    on="Vehicle_ID",
    how="left")
```

```python
from scipy.stats import f_oneway
```

```python
vehicle_groups = [
    group["Booking_Value"].dropna()
    for name, group in booking_vehicle.groupby("Vehicle_Type")]
```

```python
anova_result = f_oneway(*vehicle_groups)
print("F-statistic:", anova_result.statistic)
print("p-value:", anova_result.pvalue)
```

#### Insight: Booking Value shows no statistically significant difference across Vehicle Types (F=0.64, p=0.699 > 0.05), so vehicle type doesn't influence booking value.

## T-test:

```python
booking_calendar = bookings.merge(
    calendar,
    on="Calendar_ID",
    how="left")
```

```python
weekday_booking_value = booking_calendar.loc[
    booking_calendar["Weekend_Flag"] == False,
    "Booking_Value"].dropna()
```

```python
weekend_booking_value = booking_calendar.loc[
    booking_calendar["Weekend_Flag"] == True,
    "Booking_Value"].dropna()
```

```python
from scipy.stats import ttest_ind
```

```python
t_test_result = ttest_ind(
    weekday_booking_value,
    weekend_booking_value,
    equal_var=False)
```

```python
print("T-statistic:", t_test_result.statistic)
print("p-value:", t_test_result.pvalue)
```

#### Insight: Weekday and weekend Booking Values differ significantly (t=-66.46, p≈0.0), with weekday bookings significantly lower than weekend bookings on average.

## Chi-Square Test:

```python
booking_payment = bookings.merge(
    payments,
    on="Payment_ID",
    how="left")
```

```python
contingency_table = pd.crosstab(
    booking_payment["Booking_Status"],
    booking_payment["Payment_Method"])
```

```python
contingency_table
```

```python
from scipy.stats import chi2_contingency
```

```python
chi2, p_value, degrees_of_freedom, expected = chi2_contingency(
    contingency_table)
```

```python
print("Chi-square statistic:", chi2)
print("p-value:", p_value)
print("Degrees of freedom:", degrees_of_freedom)
```

#### Insight: Chi-square test confirms Booking Status is independent of Payment Method (p=0.924); completion rates are consistent across all payment types.
